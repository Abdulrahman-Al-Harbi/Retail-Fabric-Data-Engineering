```python
# ==========================================================
# Import Libraries
# ==========================================================

from pyspark.sql import functions as F
```

```python
# ==========================================================
# Bronze Tables
# ==========================================================

TABLES = [

    "bronze_customers",

    "bronze_products",

    "bronze_stores",

    "bronze_employees",

    "bronze_returns",

    "bronze_sales"

]
```

```python
# ==========================================================
# Row Counts
# ==========================================================

for table in TABLES:

    count = spark.table(table).count()

    print(f"{table:<25} {count:,}")
```

```python
# ==========================================================
# Schema
# ==========================================================

for table in TABLES:

    print("="*70)

    print(table)

    spark.table(table).printSchema()
```

```python
# ==========================================================
# Missing Values
# ==========================================================

def missing_values(table_name):

    df = spark.table(table_name)

    expr = [

        F.count(

            F.when(

                F.col(c).isNull(),

                c

            )

        ).alias(c)

        for c in df.columns

    ]

    display(

        df.select(expr)

    )
```

```python
missing_values("bronze_sales")
```

```python
missing_values("bronze_customers")
```

```python
# ==========================================================
# Duplicate Records
# ==========================================================

sales = spark.table("bronze_sales")

duplicates = (

    sales

    .groupBy(

        sales.columns

    )

    .count()

    .filter(

        F.col("count") > 1

    )

)

duplicates.count()
```
