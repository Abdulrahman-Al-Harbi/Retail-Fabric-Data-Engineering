```python
# ==========================================================
# Generate Customers
# ==========================================================

def generate_customers():

    loyalty_levels = [
        "Bronze",
        "Silver",
        "Gold",
        "Platinum"
    ]

    registration_channels = [
        "Website",
        "Mobile App",
        "Store"
    ]

    rows = []

    for customer_id in range(1, NUM_CUSTOMERS + 1):

        city = random.choice(list(CITY_REGION.keys()))

        birth_date = fake.date_between(
            start_date="-65y",
            end_date="-18y"
        )

        join_date = fake.date_between(
            start_date="-5y",
            end_date="today"
        )

        rows.append({

            "CustomerID": customer_id,

            "FirstName": fake.first_name(),

            "LastName": fake.last_name(),

            "Gender": random.choice([
                "Male",
                "Female"
            ]),

            "BirthDate": birth_date,

            "Email": fake.email(),

            "Phone": fake.msisdn()[:10],

            "City": city,

            "Region": CITY_REGION[city],

            "Country": "Saudi Arabia",

            "LoyaltyLevel": random.choice(
                loyalty_levels
            ),

            "RegistrationChannel": random.choice(
                registration_channels
            ),

            "EmailVerified": random.choice([
                True,
                True,
                True,
                False
            ]),

            "IsActive": random.choice([
                True,
                True,
                True,
                True,
                False
            ]),

            "JoinDate": join_date

        })

    customers = pd.DataFrame(rows)

    customers.to_csv(
        DATA_PATH / "customers.csv",
        index=False
    )

    return customers
```

```python
# ==========================================================
# Generate Products
# ==========================================================

def generate_products():

    suppliers = [
        "TechSource",
        "Prime Distribution",
        "Global Supplies",
        "Smart Retail",
        "Future Electronics"
    ]

    smartphone_storage = [
        "128GB",
        "256GB",
        "512GB",
        "1TB"
    ]

    laptop_memory = [
        "8GB RAM",
        "16GB RAM",
        "32GB RAM"
    ]

    tv_sizes = [
        '55"',
        '65"',
        '75"',
        '85"'
    ]

    monitor_sizes = [
        '24"',
        '27"',
        '32"'
    ]

    colors = [
        "Black",
        "White",
        "Silver",
        "Blue",
        "Gray"
    ]

    rows = []

    product_id = 1

    for sub_category, brands in PRODUCT_MASTER.items():

        for brand, products in brands.items():

            for product in products:

                variants = []

                # Smartphones
                if sub_category == "Smartphone":

                    variants = [

                        f"{product} {storage} {color}"

                        for storage in smartphone_storage

                        for color in colors

                    ]

                # Laptops
                elif sub_category == "Laptop":

                    variants = [

                        f"{product} {memory}"

                        for memory in laptop_memory

                    ]

                # TVs
                elif sub_category == "TV":

                    variants = [

                        f"{product} {size}"

                        for size in tv_sizes

                    ]

                # Monitors
                elif sub_category == "Monitor":

                    variants = [

                        f"{product} {size}"

                        for size in monitor_sizes

                    ]

                # Everything else
                else:

                    variants = [product]

                # Create rows
                for variant in variants:

                    if product_id > NUM_PRODUCTS:

                        break

                    min_price, max_price = PRICE_RANGE[sub_category]

                    cost = random.randint(
                        int(min_price * 0.70),
                        int(max_price * 0.80)
                    )

                    price = round(
                        cost * random.uniform(1.20, 1.40),
                        2
                    )

                    rows.append({

                        "ProductID": product_id,

                        "ProductName": variant,

                        "Category": (
                            "Electronics"
                            if sub_category in [
                                "Smartphone",
                                "Laptop",
                                "Tablet",
                                "TV",
                                "Monitor",
                                "Gaming",
                                "Headphones"
                            ]
                            else (
                                "Home Appliances"
                                if sub_category == "Kitchen"
                                else "Furniture"
                            )
                        ),

                        "SubCategory": sub_category,

                        "Brand": brand,

                        "Supplier": random.choice(suppliers),

                        "Cost": cost,

                        "UnitPrice": price,

                        "StockQuantity": random.randint(
                            20,
                            500
                        )

                    })

                    product_id += 1

                if product_id > NUM_PRODUCTS:
                    break

            if product_id > NUM_PRODUCTS:
                break

        if product_id > NUM_PRODUCTS:
            break

    products = pd.DataFrame(rows)

    products.to_csv(
        DATA_PATH / "products.csv",
        index=False
    )

    return products
```

```python
# ==========================================================
# Generate Employees
# ==========================================================

def generate_employees():

    positions = {

        "Management":[
            "Store Manager",
            "Supervisor"
        ],

        "Sales":[
            "Sales Associate",
            "Senior Sales Associate"
        ],

        "Operations":[
            "Cashier",
            "Inventory Clerk"
        ]

    }

    salary_ranges = {

        "Store Manager":(18000,25000),

        "Supervisor":(12000,17000),

        "Senior Sales Associate":(7000,10000),

        "Sales Associate":(5000,7000),

        "Cashier":(4500,6500),

        "Inventory Clerk":(5000,7000)

    }

    rows = []

    employee_id = 1

    for _, store in stores.iterrows():

        # Every store has between 10 and 18 employees
        employee_count = random.randint(10,18)

        # Always create one manager
        manager_salary = random.randint(
            salary_ranges["Store Manager"][0],
            salary_ranges["Store Manager"][1]
        )

        rows.append({

            "EmployeeID": employee_id,

            "EmployeeName": fake.name(),

            "StoreID": store["StoreID"],

            "Department": "Management",

            "Position": "Store Manager",

            "HireDate": fake.date_between("-12y","today"),

            "Salary": manager_salary

        })

        employee_id += 1

        # Remaining employees
        for _ in range(employee_count - 1):

            department = random.choice(list(positions.keys()))

            # لا نكرر المدير
            available_positions = [
                p for p in positions[department]
                if p != "Store Manager"
            ]

            position = random.choice(available_positions)

            salary = random.randint(
                salary_ranges[position][0],
                salary_ranges[position][1]
            )

            rows.append({

                "EmployeeID": employee_id,

                "EmployeeName": fake.name(),

                "StoreID": store["StoreID"],

                "Department": department,

                "Position": position,

                "HireDate": fake.date_between("-10y","today"),

                "Salary": salary

            })

            employee_id += 1

    employees = pd.DataFrame(rows)

    employees.to_csv(
        DATA_PATH / "employees.csv",
        index=False
    )

    return employees
```

```python
# ==========================================================
# Sales Configuration
# ==========================================================

SALES_PLAN = {

    2022: 80000,

    2023: 110000,

    2024: 140000,

    2025: 170000

}
```

```python
# ==========================================================
# Monthly Weights
# ==========================================================

MONTH_WEIGHTS = {

    1:0.06,
    2:0.06,

    3:0.09,
    4:0.09,

    5:0.07,
    6:0.07,
    7:0.07,
    8:0.07,

    9:0.08,
    10:0.09,

    11:0.14,

    12:0.11

}
```

```python
# ==========================================================
# Monthly Sales Calculator
# ==========================================================

def monthly_sales(total_sales):

    result = {}

    remaining = total_sales

    for month, weight in MONTH_WEIGHTS.items():

        count = int(total_sales * weight)

        result[month] = count

        remaining -= count

    result[12] += remaining

    return result
```

```python
# ==========================================================
# Lookups
# ==========================================================

product_lookup = products.set_index("ProductID")

employee_lookup = employees.set_index("EmployeeID")

customer_lookup = customers.set_index("CustomerID")

store_lookup = stores.set_index("StoreID")
```

```python
# ==========================================================
# Assign Store
# ==========================================================

def assign_store(customer_id, employee_id, sales_channel):

    # Online Orders
    if sales_channel == "Online":

        return employee_lookup.loc[
            employee_id,
            "StoreID"
        ]

    # Customer City
    city = customer_lookup.loc[
        customer_id,
        "City"
    ]

    # Stores in same city
    city_stores = stores_by_city.get(city)

    if city_stores:

        return random.choice(city_stores)

    # Fallback
    return employee_lookup.loc[
        employee_id,
        "StoreID"
    ]
```

```python
# ==========================================================
# Payment Methods
# ==========================================================

PAYMENT_METHODS = [

    "Cash",

    "Mada",

    "Visa",

    "Mastercard",

    "Apple Pay",

    "STC Pay"

]

ORDER_STATUS = [

    "Completed",

    "Completed",

    "Completed",

    "Completed",

    "Pending",

    "Returned",

    "Cancelled"

]
```

```python
# ==========================================================
# Generate One Month Sales
# ==========================================================

def generate_month_sales(year, month, num_sales, start_sale_id):

    # -----------------------------
    # Random Foreign Keys
    # -----------------------------
    customer_ids = np.random.choice(
        customers["CustomerID"].values,
        size=num_sales
    )

    product_ids = np.random.choice(
        products["ProductID"].values,
        size=num_sales
    )

    employee_ids = np.random.choice(
        employees["EmployeeID"].values,
        size=num_sales
    )

    quantities = np.random.randint(
        1,
        6,
        size=num_sales
    )

    discounts = np.random.choice(
        [0, 5, 10, 15, 20],
        size=num_sales,
        p=[0.50, 0.20, 0.15, 0.10, 0.05]
    )

    payment_methods = np.random.choice(
        PAYMENT_METHODS,
        size=num_sales
    )

    statuses = np.random.choice(
        ORDER_STATUS,
        size=num_sales
    )

    sales_channels = np.random.choice(
        ["Store", "Online"],
        size=num_sales,
        p=[0.70, 0.30]
    )

    # -----------------------------
    # Order Dates
    # -----------------------------
    start_date = pd.Timestamp(year, month, 1)
    end_date = start_date + pd.offsets.MonthEnd()

    order_dates = np.random.choice(
        pd.date_range(start_date, end_date),
        size=num_sales
    )

    # -----------------------------
    # Create DataFrame
    # -----------------------------
    sales = pd.DataFrame({

        "SaleID": np.arange(
            start_sale_id,
            start_sale_id + num_sales
        ),

        "OrderDate": pd.to_datetime(order_dates),

        "CustomerID": customer_ids,

        "ProductID": product_ids,

        "EmployeeID": employee_ids,

        "Quantity": quantities,

        "Discount": discounts,

        "PaymentMethod": payment_methods,

        "SalesChannel": sales_channels,

        "OrderStatus": statuses

    })
    sales["StoreID"] = sales.apply(

    lambda row:

    assign_store(

        row["CustomerID"],

        row["EmployeeID"],

        row["SalesChannel"]

    ),

    axis=1
    )

    return sales
```

```python
# ==========================================================
# Enrich Sales
# ==========================================================

def enrich_sales(sales):

    # Product information
    sales["UnitPrice"] = (
        product_lookup
        .loc[sales["ProductID"], "UnitPrice"]
        .values
    )

    sales["UnitCost"] = (
        product_lookup
        .loc[sales["ProductID"], "Cost"]
        .values
    )

    # Customer city
    sales["CustomerCity"] = (
        customer_lookup
        .loc[sales["CustomerID"], "City"]
        .values
    )

    # Store city
    sales["StoreCity"] = (
        store_lookup
        .loc[sales["StoreID"], "City"]
        .values
    )

    return sales
```

```python
# ==========================================================
# Business Metrics
# ==========================================================

def calculate_metrics(sales):

    sales["Revenue"] = (

        sales["Quantity"]

        * sales["UnitPrice"]

        * (1 - sales["Discount"] / 100)

    ).round(2)

    sales["Cost"] = (

        sales["Quantity"]

        * sales["UnitCost"]

    ).round(2)

    sales["Profit"] = (

        sales["Revenue"]

        - sales["Cost"]

    ).round(2)

    sales["Tax"] = (

        sales["Revenue"]

        * 0.15

    ).round(2)

    return sales
```

```python
# ==========================================================
# Create Sales Folder
# ==========================================================

SALES_PATH = DATA_PATH / "sales"

SALES_PATH.mkdir(
    exist_ok=True
)
```

```python
# ==========================================================
# Generate All Sales Files
# ==========================================================

def generate_all_sales():

    all_sales = []

    sale_id = 1

    for year, yearly_total in SALES_PLAN.items():

        monthly_plan = monthly_sales(yearly_total)

        for month, month_total in monthly_plan.items():

            print(
                f"Generating {year}-{month:02d} ({month_total:,} rows)"
            )

            sales = generate_month_sales(

                year=year,

                month=month,

                num_sales=month_total,

                start_sale_id=sale_id

            )

            sales = enrich_sales(sales)

            sales = calculate_metrics(sales)

            filename = SALES_PATH / f"sales_{year}_{month:02d}.csv"

            sales.to_csv(

                filename,

                index=False

            )

            all_sales.append(sales)

            sale_id += month_total

    all_sales = pd.concat(

        all_sales,

        ignore_index=True

    )

    return all_sales
```

```python
# ==========================================================
# Add Shipping Information
# ==========================================================

def add_shipping(sales):

    sales = sales.copy()

    sales["OrderDate"] = pd.to_datetime(
        sales["OrderDate"]
    )

    sales["ShipDate"] = pd.NaT

    sales["DeliveryDate"] = pd.NaT

    sales["DeliveryDays"] = pd.Series(
        pd.NA,
        index=sales.index,
        dtype="Int64"
    )

    shipped = sales["OrderStatus"].isin(
        [
            "Completed",
            "Returned"
        ]
    )

    shipping_days = np.random.randint(
        1,
        4,
        shipped.sum()
    )

    delivery_days = np.random.randint(
        2,
        8,
        shipped.sum()
    )

    sales.loc[
        shipped,
        "ShipDate"
    ] = (

        sales.loc[
            shipped,
            "OrderDate"
        ]

        +

        pd.to_timedelta(
            shipping_days,
            unit="D"
        )

    )

    sales.loc[
        shipped,
        "DeliveryDate"
    ] = (

        sales.loc[
            shipped,
            "ShipDate"
        ]

        +

        pd.to_timedelta(
            delivery_days,
            unit="D"
        )

    )

    sales.loc[
        shipped,
        "DeliveryDays"
    ] = (

        sales.loc[
            shipped,
            "DeliveryDate"
        ]

        -

        sales.loc[
            shipped,
            "OrderDate"
        ]

    ).dt.days

    return sales
```

```python
# ==========================================================
# Generate Returns
# ==========================================================

def generate_returns(sales):

    reasons = [

        "Damaged Product",

        "Wrong Item",

        "Changed Mind",

        "Late Delivery",

        "Defective Product"

    ]

    returned_sales = sales[

        sales["OrderStatus"] == "Returned"

    ].copy()

    rows = []

    return_id = 1

    for _, sale in returned_sales.iterrows():

        rows.append({

            "ReturnID": return_id,

            "SaleID": sale["SaleID"],

            "ReturnDate": (

                sale["DeliveryDate"]

                +

                pd.to_timedelta(

                    random.randint(1,14),

                    unit="D"

                )

            ),

            "Reason": random.choice(
                reasons
            ),

            "RefundAmount": round(

                sale["Revenue"]

                +

                sale["Tax"],

                2

            )

        })

        return_id += 1

    returns = pd.DataFrame(rows)

    returns.to_csv(

        DATA_PATH /

        "returns.csv",

        index=False

    )

    return returns
```

```python
# ==========================================================
# Inject Missing Values
# ==========================================================

def inject_missing_values(sales):

    sales = sales.copy()

    # Missing Payment Method (1%)
    payment_idx = sales.sample(
        frac=0.01,
        random_state=42
    ).index

    sales.loc[
        payment_idx,
        "PaymentMethod"
    ] = pd.NA

    # Missing Discount (0.5%)
    discount_idx = sales.sample(
        frac=0.005,
        random_state=100
    ).index

    sales.loc[
        discount_idx,
        "Discount"
    ] = pd.NA

    return sales
```

```python
# ==========================================================
# Inject Duplicate Rows
# ==========================================================

def inject_duplicates(df, duplicate_rate=0.01):

    df = df.copy()

    n_duplicates = max(
        1,
        int(len(df) * duplicate_rate)
    )

    duplicates = df.sample(
        n=n_duplicates,
        random_state=42
    )

    df = pd.concat(
        [df, duplicates],
        ignore_index=True
    )

    return df
```

```python
# ==========================================================
# Inject Customer Text Issues
# ==========================================================

def inject_customer_text_issues(customers):

    customers = customers.copy()

    # lowercase
    idx = customers.sample(
        frac=0.03,
        random_state=42
    ).index

    customers.loc[
        idx,
        "City"
    ] = customers.loc[
        idx,
        "City"
    ].str.lower()

    # UPPERCASE
    idx = customers.sample(
        frac=0.03,
        random_state=21
    ).index

    customers.loc[
        idx,
        "City"
    ] = customers.loc[
        idx,
        "City"
    ].str.upper()

    # Leading & Trailing Spaces
    idx = customers.sample(
        frac=0.03,
        random_state=8
    ).index

    customers.loc[
        idx,
        "City"
    ] = (
        " "
        + customers.loc[idx, "City"]
        + " "
    )

    return customers
```
