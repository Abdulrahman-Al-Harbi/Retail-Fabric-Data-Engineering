```python
spark.sql("""

CREATE OR REPLACE VIEW vw_sales_analysis AS

SELECT

    f.SaleID,

    d.Date,

    d.Year,

    d.Quarter,

    d.Month,

    d.MonthName,

    c.CustomerID,

    c.FirstName,

    c.LastName,

    c.City,

    c.Region,

    c.LoyaltyLevel,

    p.ProductID,

    p.ProductName,

    p.Category,

    p.SubCategory,

    p.Brand,

    s.StoreID,

    s.StoreName,

    s.City AS StoreCity,

    e.EmployeeID,

    e.EmployeeName,

    f.Quantity,

    f.Discount,

    f.Revenue,

    f.Cost,

    f.Profit,

    f.Tax,

    f.PaymentMethod,

    f.SalesChannel,

    f.OrderStatus

FROM fact_sales f

LEFT JOIN dim_customer c

ON f.CustomerID = c.CustomerID

LEFT JOIN dim_product p

ON f.ProductID = p.ProductID

LEFT JOIN dim_store s

ON f.StoreID = s.StoreID

LEFT JOIN dim_employee e

ON f.EmployeeID = e.EmployeeID

LEFT JOIN dim_date d

ON f.DateKey = d.DateKey

""")
```

```python
display(

    spark.sql(

        "SELECT * FROM vw_sales_analysis LIMIT 20"

    )

)
```
